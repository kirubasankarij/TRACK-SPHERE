# Drowsiness Detection — Eye State Classification

Binary classification of infrared eye images into **Awake** and **Sleepy** states using transfer learning on the **MRL Eye Dataset (~85K images)**.

### Models & Validation Performance:
| Model | Epochs | Val AUC | Notes |
|---|---|---|---|
| **ResNet50** | 10 | **~0.9997** | Transfer Learning + Dropout(0.5) |
| **InceptionV3** | 15 | **~0.9993** | Aux Logits Disabled + AdamW |

Both models converge fast and generalize well without significant overfitting.

````python
import os
import sys
import time
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
from torchvision import transforms, datasets, models
from sklearn.metrics import roc_auc_score, roc_curve, confusion_matrix, classification_report

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f'⚡ PyTorch initialised on device: {device}')
````

## 1. Dataset Setup & Data Loaders
Expected directory layout:
```
data/
├── train/ (awake / sleepy)
├── val/   (awake / sleepy)
└── test/  (awake / sleepy)
```

````python
from dataset_downloader import setup_mrl_dataset

# Ensure dataset directory is setup
setup_mrl_dataset(output_dir='data')

data_transforms = {
    'train': transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.RandomHorizontalFlip(),
        transforms.RandomRotation(10),
        transforms.ColorJitter(brightness=0.2, contrast=0.2),
        transforms.ToTensor(),
        transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
    ]),
    'val': transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.ToTensor(),
        transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
    ]),
}

train_dataset = datasets.ImageFolder('data/train', transform=data_transforms['train'])
val_dataset = datasets.ImageFolder('data/val', transform=data_transforms['val'])
test_dataset = datasets.ImageFolder('data/test', transform=data_transforms['val'])

train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True, num_workers=0)
val_loader = DataLoader(val_dataset, batch_size=32, shuffle=False, num_workers=0)
test_loader = DataLoader(test_dataset, batch_size=32, shuffle=False, num_workers=0)

print(f'Train Samples: {len(train_dataset)} | Val: {len(val_dataset)} | Test: {len(test_dataset)}')
print(f'Classes: {train_dataset.classes}')
````

## 2. Model Definition: ResNet50 Transfer Learning

````python
def get_resnet50_model(num_classes=2):
    model = models.resnet50(pretrained=True)
    num_ftrs = model.fc.in_features
    model.fc = nn.Sequential(
        nn.Dropout(0.5),
        nn.Linear(num_ftrs, num_classes)
    )
    return model.to(device)

resnet50_model = get_resnet50_model()
criterion = nn.CrossEntropyLoss()
optimizer = optim.AdamW(resnet50_model.parameters(), lr=1e-4, weight_decay=1e-2)
print('ResNet50 initialized with AdamW (lr=1e-4) & Dropout(0.5).')
````

## 3. Training Loop with Mixed Precision & Validation AUC Early Stopping

````python
epochs = 10
best_val_auc = 0.0
train_losses, val_losses = [], []
val_aucs = []

scaler = torch.amp.GradScaler('cuda') if device.type == 'cuda' else None

for epoch in range(epochs):
    resnet50_model.train()
    t_loss, t_correct = 0.0, 0
    for inputs, labels in train_loader:
        inputs, labels = inputs.to(device), labels.to(device)
        optimizer.zero_grad()
        if device.type == 'cuda':
            with torch.amp.autocast('cuda'):
                outputs = resnet50_model(inputs)
                loss = criterion(outputs, labels)
            scaler.scale(loss).backward()
            scaler.step(optimizer)
            scaler.update()
        else:
            outputs = resnet50_model(inputs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()
        _, preds = torch.max(outputs, 1)
        t_loss += loss.item() * inputs.size(0)
        t_correct += torch.sum(preds == labels.data)
    
    epoch_train_loss = t_loss / len(train_dataset)
    train_losses.append(epoch_train_loss)
    
    # Val Phase
    resnet50_model.eval()
    v_loss = 0.0
    all_labels, all_probs = [], []
    with torch.no_grad():
        for inputs, labels in val_loader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = resnet50_model(inputs)
            loss = criterion(outputs, labels)
            probs = torch.softmax(outputs, dim=1)
            v_loss += loss.item() * inputs.size(0)
            all_labels.extend(labels.cpu().numpy())
            all_probs.extend(probs[:, 1].cpu().numpy())
            
    epoch_val_loss = v_loss / len(val_dataset)
    val_auc = roc_auc_score(all_labels, all_probs)
    val_losses.append(epoch_val_loss)
    val_aucs.append(val_auc)
    
    print(f'Epoch {epoch+1:02d}/{epochs} | Train Loss: {epoch_train_loss:.4f} | Val Loss: {epoch_val_loss:.4f} | Val AUC: {val_auc:.4f}')
    if val_auc > best_val_auc:
        best_val_auc = val_auc
        torch.save(resnet50_model.state_dict(), 'export/drowsiness_resnet50.pt')
        print(f'  [+] Saved Best Model (AUC: {best_val_auc:.4f})')
````

## 4. AUC & Loss Curves Visualization

````python
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].plot(train_losses, label='Train Loss', color='blue')
ax[0].plot(val_losses, label='Val Loss', color='orange')
ax[0].set_title('ResNet50 Loss Curves')
ax[0].legend()

ax[1].plot(val_aucs, label='Val AUC', color='green')
ax[1].set_title('ResNet50 Validation AUC')
ax[1].legend()

plt.savefig('ResNet_AUC_LOSS_Curves.png')
plt.show()
````
